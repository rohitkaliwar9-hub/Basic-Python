# string methods

```python
#upper()
text1="hello";

print("upper :",text1.upper());
```

```python
#lower
text2="hello";

print("lower :",text2.lower());
```

```python
#title
text3="world hello";

print("title() :",text3.title());
```

```python
#swapcase swap the case
text4="hello";
print("swapcase():",text4.swapcase());
```

```python
#split
text5="a b c";
print("split():",text5.split());
```

```python
#replace
text6="apple";
print("replace():",text6.replace("a","b"));
```

```python
#strip
text7="hello";
print("strip():",text7.strip());
```

```python
#startswith
text8="python";
print("startswith():",text8.startswith("py"));
```

```python
#endswith
text9="python";
print("endswith():",text9.endswith("on"));
```

```python
#isAlpha
text10a="abc"
text10b="abc123"
print("isalpha():",text10a.isalpha())
print("isalpha():",text10b.isalpha())
```

```python
#isNumber
text11a="123"
text11b="12a"
print("isnumberic():",text11a.isnumeric())
print("isnumberic():",text11b.isnumeric())
```

```python
#isalnum
text12a="abc123"
text12b="abc 123"
print("isalnum:",text12a.isalnum())
print("isalnum:",text12b.isalnum())
```

```python
#islower
text13="hello"
print("islower:",text13.islower())
```

```python
#is upper
text14="hello"
print("isupper:",text13.isupper())
```

```python
#find
text15="apple"
print("find('p'):",text15.find('p'))
print("find('z'):",text15.find('z'))
```

# Mini realworld data

```python
name=input("enter your name:").strip();
age=input("enter your age:").strip();
email=input("enter your email:").strip();

#name validation 
if name.isalpha():
    print("name:valid")
else:
    print("name:invalid")

#age validation
if age.isnumeric():
    print("age:valid")
else:
    print("age:invalid")

#email validation
if"@" in email and "." in email:
    print("email:valid")
else:
    print("email:invalid")
```
