```python
a=int(input("Enter your Number"))
n= int(input("up to"))
for i in range (1,n+1):
 print(f" {a} * {i} = {a*i}")
```

```python
row = 5;
for i in range(1,row + 1):
    print("*"*i);
```

```python
star = 5;
for i in range(star,0,-1):
    print("*"*i);
```

```python
for i in range (6):
    for j in range (i):
        print("*",end="");
        print("")
```
